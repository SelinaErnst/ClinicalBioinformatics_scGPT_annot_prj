# UMAP

```python
# create neighborhood graph (.uns['neighbors'], .obsp['distances'] .obsp['connectivities'])
sc.pp.neighbors(adata,n_neighbors=15, knn=True, key_added="SE_neighbors")
# cluster cells based on neighborhood graph
sc.tl.leiden(adata, flavor="igraph", n_iterations=2, use_weights=True, neighbors_key="SE_neighbors", key_added="SE_leiden")
# create 2D presentation
sc.tl.umap(adata,key_added="SE_umap")
```
